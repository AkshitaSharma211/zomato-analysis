```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('zomato.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df['rate'] = df['rate'].str.replace('/5', '').str.strip()
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
```

```python
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].str.replace(',', '').str.strip()
df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce')
```

```python
df
```

```python
df['location'] = df['location'].str.strip().str.title()
df['cuisines'] = df['cuisines'].str.strip().str.title()
df['rest_type'] = df['rest_type'].str.strip().str.title()
```

```python
df = df.rename(columns={
    'approx_cost(for two people)': 'cost_for_two',
    'listed_in(type)': 'listed_type',
    'listed_in(city)': 'listed_city'
})
```

```python
import matplotlib.pyplot as plt
df['location'].value_counts().head(10).plot(kind='bar',figsize=(10,6)) 
plt.title("Top 10 Most Locations in Bangalore")
plt.xlabel("Locations")
plt.ylabel("Number of restraunts")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

df['rate'].plot(kind='hist',figsize=(10,6))
plt.title("Rating Distribution")
plt.xlabel("Ratings")
plt.ylabel("Number of restraunts")
plt.tight_layout()
plt.show()

```
