```python
data = np.array([23, 45, 12, 67, 34, 89, 56])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Std Dev:", np.std(data))
print("Variance:", np.var(data))
```

```python
marks = np.array([55, 40, 78, 33, 90, 61])
passed = marks[marks >= 50]
failed = marks[marks < 50]

print("Passed:", passed)
print("Failed:", failed)
print("Pass percentage:", len(passed)/len(marks)*100, "%")
```

```python
# Why NumPy over Python lists?
list1 = list(range(1000000))
arr1 = np.arange(1000000)

import time

start = time.time()
result = [x*2 for x in list1]
print("List time:", time.time() - start)

start = time.time()
result = arr1 * 2
print("NumPy time:", time.time() - start)
```

```python
# Normalization - used in every ML model
data = np.array([100, 200, 300, 400, 500])

normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print("Normalized:", normalized)
```

```python
# Correlation - finding relationship between variables
height = np.array([150, 160, 170, 180, 190])
weight = np.array([50, 60, 70, 80, 90])

correlation = np.corrcoef(height, weight)
print("Correlation matrix:\n", correlation)
print("Correlation value:", correlation[0][1])
# closer to 1 = strong positive relationship
# closer to -1 = strong negative relationship
# closer to 0 = no relationship
```

```python
# In ML we split data into training and testing
data = np.arange(1, 101)   # 100 samples

# Shuffle first
np.random.seed(42)
np.random.shuffle(data)

# 80% train, 20% test
split = int(0.8 * len(data))
train = data[:split]
test = data[split:]

print("Train size:", len(train))
print("Test size:", len(test))
print("Train data:", train[:5])
print("Test data:", test[:5])
```
